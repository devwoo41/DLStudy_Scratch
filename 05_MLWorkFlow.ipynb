## 머신러닝 워크플로(작업순서)

- 작업 정의

- 모델 개발

- 배포

### 작업 정의

- 문제 영역과 고객의 요청 이면에 있는 비즈니스 로직을 이해한다. 데이터를 수집하고, 데이터가 드러내는 내용을 이해하고, 작업의 성공을 측정하는 방법을 선택한다.

#### 문제 정의

```
1. 입력 데이터는 무엇인가?
2. 어떤 것을 예측하려고 하는가?
- ex. 예를 들어, 영화 리뷰와 감성 레이블이 태깅되어 있어야 영화 리뷰의 감성 분류를 학습할 수 있다.
3. 당명한 문제의 종류
- ex. 이진 분류, 다중 분류, 스칼라 회귀 등, 벡터 회귀, 다중 레이블 다중 분류...
4. 기존 솔류션
- ex. 스팸 필터링이나 신용 카드 부정 거래 감지를 처리하기 위해 기존에는 중첩된 많은 if문으로 구성된 수동으로 만든 알고리즘을 사용했을 것이다. 
5. 고려해야 할 특별한 제약
- ex. 스팸 감지 시스템을 적용할 앱은 엔드-투-엔드로 엄격히 암호화 되어 있어 스팸감지모델이 사용자의 핸드폰에 설치되어야 하고 외부 데이터셋으로 훈련되어야 한다. 

#### 데이터 수집

- 작업 특성을 이해하고, 입력과 타깃이 무엇인지 알게 되면 데이터를 수집할 차례이다. 대부분의 머신러닝 프로젝트에서 가장 힘들고 시간이 많이 걸리며 비용이 많이 드는 단계이다. (이거 ㄹㅇ)
ex. 
``` 
사진 검색 엔진 프로젝트에서는 먼저 분류하려는 일련의 레이블을 선택해야 한다. 예를 들어, 1만 개의 이미지 카테고리를 지정한다. 그 다음, 이 레이블 집합을 사용해서 사용자가 업로드한 수십만 개의 이미지에 수동으로 태깅해야 한다.

채팅 앱의 스팸 감지 프로젝트에서 사용자 대화가 엔드-투-엔드로 암호화 되기 때문에 이 콘텐츠를 모델 훈련에 사용할 수 없다. 필터링되지 않은 소셜 미디어 게시물로 구성된 별도의 데이터셋이 필요하고 이를 스팸, 공격적 또는 허용 가능으로 태그를 달아야 한다.

음악 추천 엔진의 경우 사용자가 '좋아요'를 표시한 기록을 사용할 수 있다. 
```
- 데이터 애너테이션 인프라에 투자하기
```
데이터 애너테이션(Data Annotaion)은, raw data에 meaningful label을 붙여주는 과정으로써, 이 과정이 타깃의 품질을 결정하며 결과적으로 모델의 품질을 결정한다. 

이에는 여러 가지 방법들이 있는데, 직접 데이터에 애너테이션을 수행하거나, 레이블을 모으기 위해 mechanical turk 같은 크라우드 소싱 플랫폼을 사용하거나, 전문적인 데이터 레이블링 회사의 서비스를 사용할 수 있다.

아웃소싱은 잠재적으로 시간과 비용을 절약할 수 있지만 통제권이 넘어가고, 미케니컬 터크의 경우 애너테이션에 잡음이 꽤나 들어갈 수도 있다.
```
- 대표성 없는 데이터 : 머신러닝 모델은 이전에 본 샘플과 비슷한 입력만 이해할 수 있다. 따라서 훈련에 사용하는 데이터가 제품 환경에 있는 데이터를 대표하는 것이 중요하다. 이와 관련된 현상으로 *개념이동*과 같은 현상이 있다. 이는 거의 모든 실전문제에서 만나게 되는 문제인데, 제품 환경에서 데이터의 속성이 시간에 따라 변할때 일어난다. 쉽게 얘기해 2011년에 수집한 IMDB 데이터셋으로 훈련한 모델은 2020년 리뷰에서 성능이 좋지 않는 등의 문제이다. 이는 특히, 신용 카드 부정거래 감지와 같은 적대적인 상황에서 특히 심각하다.


#### 데이터 이해

- 데이터를 블랙박스처럼 다루는 것은 안좋은 방법이다. 모델 훈련을 시작하기 전에 데이터를 탐색하고 시각화하여 예측 능력을 가진 특성에 대한 통찰을 얻어야 한다.

```
데이터가 이미지나 자연어 텍스트를 포함하고 있다면 몇 개의 샘플(그리고 레이블)을 직접 확인해야 한다.
데이터가 수치 특성을 포함하고 있다면 특성 값의 히스토그램을 그려서 값의 범위나 빈도를 파악하는 것이 좋다.
데이터가 위치 정보를 포함하고 있다면 지도에 그려봄으로써, 뚜렷한 패턴을 확인해보는 것도 좋다.
또, 만약 일부 샘플이 어떤 특성에 대해 누락된 값을 가지고 있다면, 데이터를 준비할 때 이를 처리해야 한다.
작업이 분류 문제라면 데이터에 있는 각 클래스의 샘플 개수를 출력해야 한다. 클래스의 샘플 개수가 거의 비슷하지 않다면 이런 불균형을 고려해야 한다.
```

#### 성공 지표 선택

- 프로젝트에서 성공하기 위해서는 먼저 성공이 무엇인가를 정의해야 한다. 

- 정확도인지, 정밀도나 재현율인지 이를 판가름하기 위한 성공의 지표가 프로젝트 전반에 걸쳐 내리는 모든 기술적 선택을 안내한다. 
ex. 클래스 분포가 균일한 분류 문제에서는 정확도와 ROC 곡선 아래의 면적인 ROC AUC가 일반적인 지표이다. 클래스 분포가 균일하지 않은 문제나 랭킹 문제, 다중 레이블 문제에는 정밀도와 재현율을 사용할 수 있다.

### 모델 개발

#### 데이터 준비

- 벡터화 : 신경망에서 모든 입력과 타깃은 일반적으로 부동 소수점 데이터로 이루어진 텐서이다. 사운드, 이미지, 텍스트 등 처리해야 할 것이 무엇이든지 먼저 텐서로 변환해야 한다. 이 단계를 *데이터 벡터화*라고 한다. 

- 값 정규화 : 주택 가격 예측의 경우, 특성들의 범위가 제각각이다. 어떤 특성은 작은 부동 소수점 값이고, 다른 특성은 매우 큰 정수 값을 가진다. 즉, 각 특성을 0~1 사이의 독립적으로 정규화하여 평균이 0이고 표준 편차가 1이 되도록 만들어야 한다.

- 일반적으로 비교적 큰 값이나 균일하지 않은 데이터를 신경망에 주입하는 것은 위험하다. 작은 값을 취해야 하고, 균일해야 한다.

- 누락된 값 처리하기 : 범주형 특성이라면 '누락된 값'이라는 의미의 새로운 범주를 만드는 것이 안전하다. 모델이 타깃에 대해 이것이 의미하는 바를 자동으로 학습한다. 수치형 특성이라면 '0'같은 임의의 값을 넣지 않도록 해야 한다.

#### 평가 방법 선택

- 홀드아웃 검증 : 데이터가 풍부할 때 사용한다. 앞서 보았지만, 홀드아웃 검증은 쉽게 말해서 한번 train/valid로 나누어 사용한다. 그러나 이러한 방식은, 랜덤해서 운에 의존하는 경향이 있다.

- k-겹 교차 검증 : 홀드아웃 검증을 사용하기에 샘플 개수가 너무 적을 때, 좋은 선택이다.

- 반복 k-겹 교차 검증 : 데이터가 적고 매우 정확한 모델 평가가 필요할 때 사용한다.
